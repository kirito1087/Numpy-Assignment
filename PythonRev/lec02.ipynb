```python
Name = " rahul KUMAR sharma "
Name = Name.strip()
Name = Name.title()

if all(ch.isalpha() or ch.isspace() for ch in Name):
    print(Name)
else:
    print("Invalid Name")
```

```python
st = " 45, 88 ,72, 91 , 60 "
lst = [int (x.strip()) for x in st.split(",")]
total = 0
for i in range(len(lst)):
    total += lst[i]
print(total)
print(total / len(lst))
```
