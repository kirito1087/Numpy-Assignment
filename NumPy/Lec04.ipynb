```python
import numpy as np
```

```python
marks = np.array([35, 68, 42, 91, 27, 55])
print(np.where(marks > 50))
print(np.where(marks > 50, 'Pass', 'Fail'))
```

```python
marks = np.array([35, 68, 42, 91, 27, 55])
grades = np.where(marks >= 75, 'A', np.where(marks >= 50, 'B', 'C'))
print(marks)
print(grades)
```

```python
marks = np.array([35, 68, 42, 91, 27, 55])
idx = np.where(marks > 50)[0]
print("Raw Return:", idx)
```

```python
marks = np.zeros((6, 4), dtype=int)

print(marks)
```

```python
marks = np.full((6, 4), 35)

print(marks)
```
