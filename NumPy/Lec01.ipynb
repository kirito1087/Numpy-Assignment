```python
# np.array() - 1D and 2D
x = np.array([2, 3, 4, 5])
print(x)
print(type(x))
```

```python
# np.arange(start, stop, step) stop not included 
print(np.arange(10, 5, -1))
print(np.arange(10))
```

```python
print(np.arange(1, 11).reshape(2, 5))
print(np.arange(1, 13).reshape(4, 3))
```

```python
# np.linespace(start, stop, num) divides in equal part of num
print(np.linspace(1, 10, 5))
print(np.linspace(0, 1, 5))
```

```python
# np.eye(n)
print(np.eye(4)) # 1. denotes principal diagonal in output matrix
print(np.eye(3,4))
print(np.eye(3,4,k=2)) # k is used for diagonal shift
```

```python
# .shape, .ndim, .size, ,dtype
x = np.arange(20).reshape(4, 5)
print(x)
print(x.shape) # shape is attribute so we dont use parenthesis 

print(x.shape)

print(x.size)
print(len(x))
print(x.ndim) # .ndim tells number of dimensions 
print(type(x))
print(x.dtype)
```

```python
arr = np.arange(4) # use reshape because .shape treats as one dimensional 
print(arr.shape)
print(arr)
```

```python
# 1D indexing - positive and negative
arr = np.arange(10)
print(arr)
print(arr[-3])
print(arr[-3 : -12 : -2])
```

```python
# 2D indexing - arr[row, col], full row, full column 
arr = np.arange(20).reshape(5, 4)
print(arr)
print(arr[2, 2])
```

```python
# 2D slicing - sub-matrix (block)
print(arr)
# print(arr[2:, 1:])
# print(arr[1, 1:])
# print(arr[::4, ::3])
print(arr[1:3:1, 2::-1])
```

```python
print(arr[1, 1::])
```
