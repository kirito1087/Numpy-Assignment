```python
# arithmetic - array with scalar, array with array
arr1 = np.arange(10)
print(arr1)
arr1 = arr1 + 5
print(arr1)
print("Multiply", arr1 * 5)
print("Exponent", arr1 ** 2)
```

```python
x = np.arange(10)
y = np.arange(11, 21)
print(x)
print(y)
print(x + y) #Length of x and y should be same for this operation to perform successfully
```

```python
x = np.arange(6).reshape(2, 3)
y = np.arange(3).reshape(1, 3)
print(x)
print(y)
print(x + y) #SIZE SAME HONA CHAHIYE 

#Now the next example will throw error
# x = np.arange(6).reshape(3, 2)
# y = np.arange(3).reshape(1, 3)
# print(x)
# print(y)
# print(x + y)
```

```python
# comparison operators - boolean arrays
x = np.arange(5)
y = np.arange(1, 6)
print(x)
print(y)
print(x == y) # Size should be same of the both the arrays 
print(x < y) 
```

```python
# np.random.seed() + np.random.rand()
print(np.random.rand(5))
print(np.random.rand(5)) # Each time we write random.rand then it will produce 5 different random values between 0 and 1
# 1 is excluded [0, 1)
np.random.seed(3) # making some internal formula static through which random values are coming for random.rand
print(np.random.rand(4)) # in other words seed is making the formula static and now if we write these two code in any computer then random values will be same 
```

```python
np.random.randn(10) # values are near to 0 such that mean is 0 and standard deviation is 1, but range is not fixed like random.rand
```

```python
# np.random.randint(low, high, size), low included and high excluded 
print(np.random.randint(10, 20, 5))
print(np.random.randint(10, 20, 5)) #everytime producing different values 
```

```python
# np.random.choice(), np.random.shuffle()
lst = [3, 4, 56, 7, 9]
print(np.random.choice(lst, 3)) # It will select 3 random values between list elements 
print(np.random.choice(lst, 3, replace = False)) # replace - false : means elemets wont repeat 
print(np.random.choice(1000, 56)) # Now it will select 56 random values between 0 and 999 
```

```python
arr = np.arange(10)
np.random.shuffle(arr) # shuffles the elemets randomly between 0 and 9
print(arr)
```
