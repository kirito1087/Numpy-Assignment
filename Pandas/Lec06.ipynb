```python
summary = tip.groupby('day').agg(
    bill_count = ('total_bill', 'size'),
    avg_bill = ('total_bill', 'mean'),
    avg_tip = ('tip', 'mean')
).round(2)
summary
```

```python
# Aggregate Function 
summary = tip.groupby("day").agg(
    bill_count = ("total_bill", 'size'),
    avg_bill = ("total_bill", "mean"),
    average_tip = ("tip", "mean")
).round(2)
summary
```

```python
summary = summary.reset_index()
summary
```

```python
summary[summary["average_tip"] >= 3]
```
