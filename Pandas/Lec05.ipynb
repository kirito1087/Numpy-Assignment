```python
import pandas as pd
import numpy as np
```

```python
titanic = pd.read_csv("../Assignment/Titanic.csv")
tip = pd.read_csv("tip.csv")

print(tip.shape, titanic.shape)
```

```python
tip.groupby('day') #returns address only
```

```python
# how many bills on each day
x = tip.groupby('day').size() # size - is used to count 

# average tip on each day
y = tip.groupby('day')['tip'].mean().round(2) # round(2) - round-off till 2 decimal places
print(x)
print()
print()
print(y)
```

```python
# make a tip_pct column, then average it per day
tip['tip_pct'] = (tip['tip'] / tip['total_bill'] * 100).round(2)
print(tip.head(2))
print(tip.groupby('day')['tip_pct'].mean().round(2))
```

```python
x = tip.groupby('day').mean(numeric_only=True)['tip']
y = tip.groupby('day')['tip'].mean()
print(x)
print(y)
```
