## PANDAS

```python
import pandas as pd
import numpy as np
```

```python
# pd.DataFrame() - creates a table from a dictionary 
restaurant_data = {
    'Table_No' : [12, 14, 15],
    'Bill_Amount' : [1250, 890, 2400],
    'Waiter_Name' : ['Amit', 'Priya', 'Rahul']
}

df = pd.DataFrame(restaurant_data)
print(df)
print(type(df))

print(df['Bill_Amount'])
print(type(df['Bill_Amount']))
```

```python
# pd.Series - creates Series (1D data with an index)
total_bill = [4700, 6744, 760, 4583]
s = pd.Series(total_bill, index = ['mon', 'tue', 'wed', 'thurs']) #index is used to change 0,1,2,.... to replaced index
print(s)
print(type(s))
```

```python
# .values / .to_numpy() - the NumPy array underneath
print(df.values)
print(type(df.values))
```

```python
#Q:
s = [45, 30, 15, 60]
s = pd.Series(s, index = ['PIZZA', 'BURGER', 'PASTA', 'FRIES'])
print(s)
```

```python
#Q
p = {
    'Brand' : ['Nike', 'Puma', 'Campus'],
    'Price' : [2500, 4000, 8000]
}
df = pd.DataFrame(p)
print(df)
```

```python
# pd.read_csv() - used to read/import a CSV file into a DataFrame
df = pd.read_csv("tip.csv")
print(df)
```
